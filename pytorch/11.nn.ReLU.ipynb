```python
import shutil

import torch
import torch.nn as nn

import torchvision.transforms as transforms
import torchvision
from torch.utils.tensorboard import SummaryWriter
from torch.utils.data import DataLoader

shutil.rmtree("../logs",ignore_errors=True)

dataset = torchvision.datasets.CIFAR10("../4Dataset",train=False,transform=transforms.ToTensor(),download=False)
dataloader = DataLoader(dataset,batch_size=16)

class Chi(nn.Module):
    def __init__(self):
        super(Chi,self).__init__()
        self.ReLU = nn.ReLU()
        
    def forward(self,x):
        x = self.ReLU(x)
        return x
    
chi = Chi()
writer = SummaryWriter("../logs")
    
    
step = 0
for data in dataloader:
    imgs, targets = data
    imgs -= 0.5
    output = chi(imgs)
    writer.add_images("input",imgs + 0.5,step)
    writer.add_images("output",output,step)
    step += 1
    
writer.close()
    
```
